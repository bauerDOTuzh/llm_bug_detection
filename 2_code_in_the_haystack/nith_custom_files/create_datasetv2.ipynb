```python

def find_subpart_of_code(required_chars, allowed_difference=150):
    attempts = 0
    chars_needed = int(required_chars+allowed_difference)
    while attempts < 5:
        # get one random file from content
        best_match = None
        best_diff = float('inf')

        for length, content in list(contents.items()):
            current_diff = abs(length - chars_needed )
            if current_diff < best_diff and length < required_chars + allowed_difference:
                best_match = content
                best_diff = current_diff

        if best_match:
            random_key = len(best_match)
            random_content = best_match

            if abs(len(random_content)-required_chars) < allowed_difference:
                del contents[random_key]  # Remove used content to avoid reuse
                return random_content
            else:
                # split the content at required chars + allowed difference
                split = random_content[0:chars_needed+1000]
                # find the last class declaration
                split_point = split.rfind('class')
                if split_point == -1:
                    attempts += 1
                else:
                    content = split[:split_point]
                    diff = len(content) - (chars_needed)
                    if diff > 0:
                        del contents[random_key]  # Remove used content to avoid reuse
                        return content
                    
                    del contents[random_key]
                    attempts += 1
        else:
            attempts += 1
    
    return None


def find_best_content(required_chars, allowed_difference=150, max_attempts=5):
    total_length_needed = required_chars
    if(required_chars <= 30):
        return ""
    
    best_match = None
    best_diff = float('inf')
    attempts = 0

    while attempts < max_attempts:
        for length, content in list(contents.items()):
            current_diff = abs(length - total_length_needed)
            if current_diff < best_diff and length < required_chars + allowed_difference:
                best_match = content
                best_diff = current_diff
        # if best_match:
            # print(f"Best match Length: {best_diff}")
            # print(f"Best match: {len(best_match)}")
        
        if best_match:
            # print(f"best match is {len(best_match)}")
            diff = required_chars - len(best_match)
            del contents[len(best_match)]  # Remove used content to avoid reuse
            
            # for cases where we need much more
            if(diff >= allowed_difference*3 and diff > 0):
                # print(f"Missing {required_chars-len(best_match)} characters")
                best_match = best_match+" \n"+ find_best_content(diff, allowed_difference, max_attempts)
            elif(diff > 0):
                #  print(f"Missing {required_chars-len(best_match)} characters")
                 part = find_subpart_of_code(diff, allowed_difference)
                 if(part):
                    # print(f"found {len(part)}")
                    best_match = best_match+" \n"+part
            return best_match
        else:
            # If no match is found within the allowed difference, regenerate contents
            attempts += 1
            print(f"No adequate match found. Regenerating contents... Attempt {attempts}")

    # After max attempts, return the content with the smallest diff even if it does not meet the allowed difference
    print("Max attempts reached. Cutting it close...")
    return find_subpart_of_code(required_chars, allowed_difference)

# Example usage
contents = regenerate_array_with_length()
required_chars = 5400  # Example required chars
content = find_best_content(required_chars, 50)
print(len(content))
print(required_chars - len(content))
```

```python
# function to create a file of a certain length and with certain posiiton
# not coherent percentage determines the split between prepend and append
def create_input(variant, target_characters, target_bug_position, p_not_coherent = 0, allowed_difference=250):
    if(target_bug_position > target_characters):
        raise Exception("Target bug position is greater than target characters")
    
    if(p_not_coherent > 0):
        raise Exception("Currently not allowed for inbetween content")
    
    content = globals()[f'{variant}']
    for i in range(0, 2):

        # take initial prompt
        buggy_line = buglines[f'{variant}']['vulnerable_line']
        # calculate current lenght
        current_length = len(content)
        # determine position in characters from start of bug (bug is "echo $this->requestHandler->getValue('output'); ")
        current_bug_position = content.find(buggy_line)

        if(target_characters < current_length):
            raise Exception("Target characters is less than current length")
        
        if(target_bug_position < current_bug_position):
            raise Exception("Target bug position is less than current bug position")
        
        characters_to_add = target_characters - current_length
        bug_characters_to_add_to_reach_position = target_bug_position - current_bug_position

        prepend_characters_total =  bug_characters_to_add_to_reach_position
        append_characters = characters_to_add - prepend_characters_total

        # update inbetween characters
        prepend_characters =  prepend_characters_total*(1-p_not_coherent)
        inbetween_characters = prepend_characters_total - prepend_characters

        
        # print all current information
        print("Current Length: ", current_length)
        print("Current Bug Position: ", current_bug_position)
        print("Characters to Add: ", characters_to_add)
        print("Bug Characters to Add to Reach Position: ", bug_characters_to_add_to_reach_position)
        print("Prepend Characters: ", prepend_characters)
        print("inbetween Charatcers: ", inbetween_characters)
        print("Append Characters: ", append_characters)

        if(append_characters < 0):
            print("they were less")
            content = globals()[f'{variant}_permutated']
        else:
            break;
    
    if append_characters < 0:
        raise Exception("Append characters is negative")

    

    # find best prepend content
    prepend_content = find_best_content(prepend_characters, int(allowed_difference/3))
    inbetween_content = find_best_content(inbetween_characters, int(allowed_difference/3))
    append_content = find_best_content(append_characters, int(allowed_difference/3))

    # print all starts
    print("prepend content len: ", len(prepend_content))
    print("inbetween content len: ", len(inbetween_content))
    print("append content len: ", len(append_content))

    # replace all variables with the content
    content = content.replace("{prepend_content}", prepend_content)
    content = content.replace("{inbetween_content}", inbetween_content)
    content = content.replace("{append_content}", append_content)

    return {
        "content": content,
        "total_length": len(content),
        "bug_pos": content.find(buggy_line)
    }
```
