```python
CWE_LIST = ["CWE-22", "CWE-89", "CWE-79"]

MODELS = ['mixtral-8x7b-32768', 'mixtral-8x22b-65536', "llama-3-70b-8192", "gpt-3.5-turbo", "gpt-4-turbo", "gpt-4o"]

model_renames = {
    "mixtral-8x7b-32768": "mixtral-8x7b",
    "mixtral-8x22b-65536": "mixtral-8x22b",
    "llama3-70b-8192": "llama-3-70b",
    "llama-3-70b-8192": "llama-3-70b"
}

max_bug_pos_in_data = 25000
threshold_step = 500

# size_threshold_type = 'bug_pos'
size_threshold_type = 'file_length_chars'
```

# Load Data

```python
real_data_df = pd.read_csv('./real_data.csv')
code_in_the_haystack_data_df = pd.read_csv('./code_in_the_haystack_data.csv')

df = pd.concat([real_data_df, code_in_the_haystack_data_df], ignore_index=True)
df['model'] = df['model'].replace('llama3-70b-8192', 'llama-3-70b-8192')
```

# Logistic Regression

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import csv
import os

# Logistic Regression Plotting Functions
def plot_logistic_regressions_for_all_models(df, models, features, cwe_ids, size_threshold, save_path='./', filename='logistic_regression_all_models'):
    
    fig, axes = plt.subplots(nrows=len(features), ncols=len(models), figsize=(8 * len(features), len(models)), dpi=100, sharex='row', sharey='all')
    fig.subplots_adjust(hspace=0.4, wspace=0.3)  # Adjust spacing between plots
    fig.suptitle(filename, fontsize=16)
    colors = plt.cm.viridis(np.linspace(0, 1, len(cwe_ids)))  # Generate a color array

    lr_stats = []
    for row, feature in enumerate(features):
        for col, model in enumerate(models):
            ax = axes[row, col] if len(features) > 1 else axes
            for cwe_id, color in zip(cwe_ids, colors):
                cwe_df = df[(df['CWE_ID'] == cwe_id) & (df['model'] == model)]
                print(f'For {cwe_id}, {model} has {len(cwe_df)} datapoints with files smaller than {size_threshold} characters.')
                if not cwe_df.empty:
                    try:
                        p_value, intercept, odds_ratios, coefficient = plot_logistic_regression(cwe_df, feature, 'classification_TP', cwe_id, ax, color)
                        lr_stats.append({
                            'CWE-ID':cwe_id, 
                            f'{size_threshold_type}_threshold':size_threshold, 
                            'model': model,
                            'p_value': p_value,
                            'coefficient': coefficient,
                            'odds_ratios': odds_ratios,
                            'intercept': intercept,
                            'feature': feature,
                        })
                    except Exception as e:
                        print('Error:', e)
            feature_label = 'Bug Position' if feature=='bug_pos' else 'File Size'
            if col == 0:  # Set Y-axis labels for the first column only
                ax.set_ylabel("Detection Probability")
            ax.set_xlabel(feature_label)
            ax.legend(loc='upper right', fontsize='small')

            # Optionally set titles for the first row, or adjust according to your needs
            if row == 0:
                # Renaming models
                ax.set_title(model_renames.get(model,model))

    #fig.suptitle(f'Logistic Regression Analysis Across CWEs for All Models', y=1.02)
    plt.tight_layout()
    plt.savefig(os.path.join(save_path,f"{filename}.pdf"))
    plt.show()
    return lr_stats

def plot_logistic_regression(data, x_col, y_col, label, ax, color):
    X = sm.add_constant(data[x_col])
    y = data[y_col]
    model = sm.Logit(y, X).fit(disp=0)
    x_values = np.linspace(data[x_col].min(), data[x_col].max(), 100)
    y_probs = model.predict(sm.add_constant(x_values))
    
    label_dict = {
        "CWE-22": "CWE-22",
        "CWE-89": "CWE-89",
        "CWE-79": "CWE-79"
    }
    
    # Define line styles for each CWE for differentiation
    line_styles = {
        'CWE-22': 'solid',  # solid line
        'CWE-89': 'dashed', # dashed line
        'CWE-79': 'dashdot' # dash-dot line
    }
    _label = label_dict.get(label, label) + (f" (p={model.pvalues[x_col]:.3f})" if model.pvalues[x_col] > 0.05 else "")
    ax.plot(x_values, y_probs, linestyle=line_styles[label], label=_label, color=color, linewidth=2)
    
    intercept = model.params['const']
    coefficient = model.params[x_col]
    odds_ratios = np.exp(intercept)
    p_value = model.pvalues[x_col]
    return p_value, intercept, odds_ratios, coefficient

lr_stats = []
for j in range(max_bug_pos_in_data//threshold_step, 0, -1):
    size_threshold = j*threshold_step
    
    df = df[df[size_threshold_type] <= size_threshold]

    # Plotting logistic regression for each model in a single plot
    lr_stats += plot_logistic_regressions_for_all_models(
        df, 
        MODELS, 
        ['file_length_chars', 'bug_pos'], 
        CWE_LIST,
        size_threshold,
        save_path='./lr_plots',
        filename=f'logistic_regression_t={size_threshold}_type={size_threshold_type}'
    )
    
# Writing to a CSV file
with open(f'./lr_stats_type={size_threshold_type}.csv', mode='w', newline='') as file:
    writer = csv.DictWriter(file, fieldnames=list(lr_stats[0].keys()))
    writer.writeheader()
    writer.writerows(lr_stats)
```
