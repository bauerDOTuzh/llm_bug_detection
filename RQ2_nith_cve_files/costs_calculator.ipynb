```python
# import vsc calles files
import pandas as pd
import tiktoken

# read the csv file
df = pd.read_csv('files.csv')
def num_tokens_from_string(string: str, encoding_name='cl100k_base') -> int:
    encoding = tiktoken.get_encoding(encoding_name)
    return len(encoding.encode(string))
```

```python
# apply to content
df['num_tokens'] = df['content'].apply(num_tokens_from_string)
```

```python
# add to each 250 in num_tokens
df['num_tokens_withPrompt'] = df['num_tokens'].apply(lambda x: x + 250)
```

```python
sum_tokens = df['num_tokens_withPrompt'].sum()
sum_tokens
```
